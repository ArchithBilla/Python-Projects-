16) Verify that every order_id in shipments_raw exists in orders_raw, confirming that all recorded shipments relate to actual orders.
 > Datasets & Columns: shipments_raw (order_id), orders_raw (order_id)


```python
def clean_order_id(col):
    col = col.str.replace('ORD-', '')
    col = col.str.replace('Ord-', '')
    col = col.str.replace('ord-', '')
    col = col.str.replace('o-', '')
    col = col.str.replace('O-', '')
    col = col.str.replace(' ', '')
    col = col.str.replace('ord', '')
    col = col.str.replace('ORD', '')
    return col
```

```python
cleann = clean_order_id(df_shipments['order_id_raw'])
```

```python
clean_order_id(df_orders['order_id_raw'])
```
