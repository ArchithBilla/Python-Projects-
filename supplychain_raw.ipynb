**10) Identify missing values in shipments_raw and count them to determine if any shipping information is incomplete.**

> shipments_raw (e.g., shipment_date, carrier, other relevant fields)

```python
missing_shipment_details = df_shipments[['ship_dt_raw', 'carrier_raw']]
missing_shipment_details
```

```python
missing_shipment_details['ship_dt_raw'] = (
    missing_shipment_details['ship_dt_raw']
    .str.replace('/', '-')
    .str.replace('Jan', '02')
    .str.replace(' ', '-')
    .str.replace(',', '-')
    .str.replace('.', '-')
)
```

```python
missing_shipment_details['ship_dt_raw']
```

```python
missing_shipment_details['ship_dt_raw'] = pd.to_datetime(missing_shipment_details['ship_dt_raw'], format= "mixed")
```

```python
missing_shipment_details['ship_dt_raw'].isna()
```

> Identified the missing [ship_dt_raw] dates rows

```python
missing_shipment_details[missing_shipment_details['ship_dt_raw'].isna()]
```

```python
missing_shipment_details['carrier_raw'] = missing_shipment_details['carrier_raw'].fillna('Unknown')
```

```python
missing_shipment_details['carrier_raw'] = (
    missing_shipment_details['carrier_raw']
    .str.replace(' ', '')
    .str.replace('fx', 'FEDEX')
    .str.replace('usps', 'USPS')
    .str.replace('ups', 'UPS')
    .str.replace('fedex', 'FEDEX')
    .astype(str)
)
```

```python
missing_shipment_details['carrier_raw']
```
